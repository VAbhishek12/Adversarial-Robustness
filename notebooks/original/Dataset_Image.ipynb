```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from torchvision import datasets, transforms

# Load CIFAR-10 test set
transform = transforms.ToTensor()
dataset = datasets.CIFAR10(root='./data', train=False, download=True, transform=transform)

classes = ['Airplane', 'Automobile', 'Bird', 'Cat', 'Deer',
           'Dog', 'Frog', 'Horse', 'Ship', 'Truck']

# Pick one clean sample per class
class_images = {}
for img, label in dataset:
    if label not in class_images:
        class_images[label] = img
    if len(class_images) == 10:
        break

# Plot — 2 rows × 5 cols
fig, axes = plt.subplots(2, 5, figsize=(10, 4.5))
fig.suptitle('CIFAR-10 Dataset — Sample Images per Class',
             fontsize=13, fontweight='bold', y=1.02)

for idx, ax in enumerate(axes.flat):
    img = class_images[idx].permute(1, 2, 0).numpy()
    ax.imshow(img)
    ax.set_title(classes[idx], fontsize=10, fontweight='bold', pad=4)
    ax.axis('off')
    # Add border
    for spine in ax.spines.values():
        spine.set_visible(True)
        spine.set_edgecolor('black')
        spine.set_linewidth(1.5)

plt.tight_layout()
plt.savefig('Fig_CIFAR10_Samples.png', dpi=300, bbox_inches='tight',
            facecolor='white')
plt.show()
print("Saved: Fig_CIFAR10_Samples.png")
```
